```python
df = df.drop(columns=['Unnamed: 0'],axis=1)
```

```python
df["Ram"] = df["Ram"].str.replace("GB", "").astype(int)
```

```python
df["Weight"] = df["Weight"].str.replace("kg", "").astype(float)
```

```python
df.head()
```

```python
df["ScreenWidth"] = df["ScreenResolution"].str.extract(r"(\d+)x\d+").astype(int)

df["ScreenHeight"] = df["ScreenResolution"].str.extract(r"\d+x(\d+)").astype(int)

df["IPS"] = df["ScreenResolution"].str.contains("IPS", case=False).astype(int)
```

```python
df[["ScreenWidth", "ScreenHeight", "IPS"]].head()
```

```python
df.drop("ScreenResolution", axis=1, inplace=True)
```

```python
df["CpuBrand"] = df["Cpu"].str.split().str[0]

df["CpuSpeed"] = df["Cpu"].str.extract(r"(\d+\.?\d*)GHz").astype(float)
```

```python
df.drop("Cpu", axis=1, inplace=True)
```

```python
import re

def get_storage(memory, storage_type):

    pattern = rf"(\d+(?:\.\d+)?)\s*(GB|TB)\s+{storage_type}"

    match = re.search(pattern, memory)

    if match:
        value = float(match.group(1))
        unit = match.group(2)

        if unit == "TB":
            value *= 1024

        return value

    return 0
```

```python
df["SSD_GB"] = df["Memory"].apply(lambda x: get_storage(x, "SSD"))
df["HDD_GB"] = df["Memory"].apply(lambda x: get_storage(x, "HDD"))
df["Hybrid_GB"] = df["Memory"].apply(lambda x: get_storage(x, "Hybrid"))
df["Flash_GB"] = df["Memory"].apply(lambda x: get_storage(x, "Flash"))
```

```python
df.head()
```

```python
df.drop("Memory", axis=1, inplace=True)
```

```python
df["GpuBrand"] = df["Gpu"].str.split().str[0]
```

```python
df["GpuBrand"].value_counts()
```

```python
df.drop("Gpu", axis=1, inplace=True)
```

```python
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
x = df.drop("Price", axis=1)
y = df["Price"]
```

```python
numerical_cols = [
    "Inches",
    "Ram",
    "Weight",
    "ScreenWidth",
    "ScreenHeight",
    "IPS",
    "CpuSpeed",
    "SSD_GB",
    "HDD_GB",
    "Hybrid_GB",
    "Flash_GB"
]
```

```python
categorical_cols = [
    "Company",
    "TypeName",
    "OpSys",
    "CpuBrand",
    "GpuBrand"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R2 Score:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ))
])

rf_model.fit(x_train, y_train)

y_pred_rf = rf_model.predict(x_test)

print("R2:", r2_score(y_test, y_pred_rf))
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", GradientBoostingRegressor(random_state=42))
])

gb_model.fit(x_train, y_train)

y_pred_gb = gb_model.predict(x_test)

print("R2:", r2_score(y_test, y_pred_gb))
```

```python
final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", GradientBoostingRegressor(random_state=42))
])
```

```python
final_model.fit(x_train, y_train)
```

```python
import pickle

with open("laptop_price_model.pkl", "wb") as f:
    pickle.dump(final_model, f)
```
